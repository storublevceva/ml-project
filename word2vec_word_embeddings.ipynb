```python
!pip install numpy datasets gensim
import numpy as np
import re
from datasets import load_dataset
from gensim.models import Word2Vec
from collections import Counter

data = load_dataset("emotion", trust_remote_code=True)
texts = data['train']['text']

def tokenize_text(text):
    return re.findall(r'\b\w+\b', text.lower())

sentences = [tokenize_text(text) for text in texts]

model = Word2Vec(
    sentences=sentences,
    vector_size=100,  # размерность текстов
    window=5,         # размер контекстного окна
    min_count=5,      # минимальную частоту слова
    workers=4,        # количество потоков
    sg=0              # использование Continuous Bag-of-Words
)

all_words = ' '.join(texts).split()
word_counts = Counter(all_words)

words = [word for word, _ in word_counts.most_common(100)]

print(words)
```

```python
target_words = ["feel", "day", "know", "myself", "really", "time", "people", "think", "life", "love"]

print("\nРезультаты")
for word in target_words:
    # Проверяем, есть ли слово в словаре модели
    if word in model.wv.key_to_index:
        # Получаем эмбеддинг
        embedding = model.wv[word]

        similar_words = model.wv.most_similar(word, topn=5)

        print(f"Слово: '{word}' (Размерность вектора: {embedding.shape})")
        for sim_word, cos_sim in similar_words:
            print(f"Ближайшее слово: '{sim_word}' | Косинусная близость: {cos_sim:.4f}")
        print()
```

```python
from sklearn.datasets import fetch_20newsgroups
categories = [
    'comp.sys.ibm.pc.hardware',
    'comp.sys.mac.hardware',
    'comp.graphics',
    'comp.windows.x'
]
data = fetch_20newsgroups(subset='all', categories=categories, shuffle=True, random_state=42)
texts = data.data

def tokenize_text(text):
    return re.findall(r'\b\w+\b', text.lower())

sentences = [tokenize_text(text) for text in texts]

model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=5,
    sg=0,
    workers=4
)

all_words = ' '.join(texts).split()
word_counts = Counter(all_words)

words = [word for word, _ in word_counts.most_common(100)]

print(words)
```

```python
target_words = ["use", "University", "only", "your", "file", "using", "anyone", "image", "drive", "program"]

print("\nРезультаты")
for word in target_words:
    if word in model.wv.key_to_index:
        embedding = model.wv[word]

        similar_words = model.wv.most_similar(word, topn=5)

        print(f"Слово: '{word}' (Размерность вектора: {embedding.shape})")
        for sim_word, cos_sim in similar_words:
            print(f"Ближайшее слово: '{sim_word}' | Косинусная близость: {cos_sim:.4f}")
        print()
```
